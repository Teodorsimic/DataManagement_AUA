### 1.3 Boolean masks and replacement — 8 points
Extract all values in units that are at least 15 and less than 30 using a Boolean mask.
Then use np.where to create adjusted_units, replacing values below 10 with 0 and keeping all other values.
Display adjusted_units - units.
Explain why & and parentheses are needed when combining the two array conditions.

```python
units[(units >= 15) & (units < 30)]
```

```python
adjusted_units = np.where(units < 10, 0, units)
```

```python
adjusted_units
```

```python
adjusted_units - units
```

Why "&": Each comparison like units >= 15 returns an array of booleans, and and raises an error because it can't decide whether a whole array is true or false; & works element-wise, comparing the two arrays position by position.

Why parentheses: Because of operator precedence, "&" is evaluated before the comparisons, so without parentheses Python would try 15 & units first; wrapping each comparison in parentheses forces them to be computed first.
